```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# Step 1: Download the webpage
url = 'https://www.macrotrends.net/stocks/charts/TSLA/tesla/revenue'
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.0.0 Safari/537.36'}  # Adding headers to mimic a web browser
response = requests.get(url, headers=headers)  # Passing the headers to the request
soup = BeautifulSoup(response.text, 'html.parser')

# Step 2: Find the table containing the quarterly revenue data
# Updated class name to reflect the current website structure
tesla_revenue_table = soup.find("table", {"class": "table"})
# Initialize an empty DataFrame to store the data
tesla_revenue = pd.DataFrame(columns=["Date", "Revenue"])

# Step 3: Extract rows and fill the DataFrame
# Check if tesla_revenue_table is not None before accessing tbody
if tesla_revenue_table:
    for row in tesla_revenue_table.tbody.find_all("tr"):
        col = row.find_all("td")
        if len(col) == 2:
            date = col[0].text.strip()
            revenue = col[1].text.strip().replace(',', '')  # Remove commas in revenue
            tesla_revenue = pd.concat([tesla_revenue, pd.DataFrame([{"Date": date, "Revenue": revenue}])], ignore_index=True) # Use concat to append data efficiently

# Step 4: Display the last five rows
print(tesla_revenue.tail())
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# Step 1: Download the webpage for GME revenue data
url = 'https://www.macrotrends.net/stocks/charts/GME/gamestop/revenue'
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.0.0 Safari/537.36'}
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, 'html.parser')

# Step 2: Find the table containing the quarterly revenue data
gme_revenue_table = soup.find("table", {"class": "table"})  # Updated class name

# Step 3: Create and populate the gme_revenue DataFrame
gme_revenue = pd.DataFrame(columns=["Date", "Revenue"])
if gme_revenue_table:
    for row in gme_revenue_table.tbody.find_all("tr"):
        col = row.find_all("td")
        if len(col) == 2:
            date = col[0].text.strip()
            revenue = col[1].text.strip().replace(',', '')
            gme_revenue = pd.concat([gme_revenue, pd.DataFrame([{"Date": date, "Revenue": revenue}])], ignore_index=True)

# Step 4: Display the last five rows of the gme_revenue DataFrame
print(gme_revenue.tail())
```

```python
import yfinance as yf
import matplotlib.pyplot as plt

# Download GameStop stock data using yfinance (for example, using 'GME' as the ticker symbol)
gme_data = yf.download("GME", start="2020-01-01", end="2023-01-01")

# Define the make_graph function
def make_graph(data, title="Stock Price Data"):
    plt.figure(figsize=(10,6))
    plt.plot(data['Close'], label='Close Price', color='blue')
    plt.title(title)
    plt.xlabel('Date')
    plt.ylabel('Price (USD)')
    plt.legend()
    plt.grid(True)
    plt.show()

# Create the graph for GameStop stock data
make_graph(gme_data, title="GameStop (GME) Stock Price")
```
